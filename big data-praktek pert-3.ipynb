```python
# data cleaning

# The original code attempted to drop columns that are not present in 'df'.
# I am commenting out or removing this line to avoid errors.
# data = data.drop(columns=['Datetime', 'Tweet_Id', 'Username', 'Label', 'Kata_kunci'])

# Using 'df' DataFrame and 'content' column for cleaning
df['content'] = df['content'].fillna('') # Changed to reassign instead of using inplace=True

import re  # Menggunakan modul re
# membuat fungsi untuk cleaning data dari kolom
def datacleaning(text):
    text = str(text) # Ensure text is a string
    text = re.sub(r'[^A-Za-z0-9\s]+', '', text)  # menghapus karakter
    text = re.sub(r'[^A-Za-z0-9]+', ' ', text)   # menghapus huruf/karakter
    text = re.sub(r'RT[\s]+', '', text)           # menghapus retweet
    text = re.sub(r'http\S+', '', text)           # menghapus URL
    text = re.sub(r'[0-9]+', '', text)            # menghapus angka
    text = text.replace('\n', ' ')                # menghapus baris baru menjadi spasi
    text = text.strip()                           # menghapus spasi dari awal dan akhir
    return text

df['content'] = df['content'].apply(datacleaning)
display(df.head())
```
